## A. Word Rotation Problem

### The function:

```python
def rotate_word(text):
    return text[1:] + text[0]

user_input = input("Enter a word to rotate")

if user_input:
    result = rotate_word(user_input)
    print(f"Rotated word: {result}")
else:
    print("Please enter a valid non-empty string.")
```

```python
#Tests for A
print(rotate_word("python"))
print(rotate_word("logic")) 
print(rotate_word("Code"))
print(rotate_word("A"))
```

## B. USERNAME BUILDER PROBLEM

### The function: 

```python
def make_username(first_name, last_name):
    first = first_name.lower().replace(" ", "")
    last = last_name.lower().replace(" ", "")
    return f"{first}.{last}"

first_name_input = input("Enter first name: ")
last_name_input = input("Enter last name: ")

if first_name_input and last_name_input:
    username = make_username(first_name_input, last_name_input)
    print(f"Generated username: {username}")
else:
    print("Please provide both first and last name.")
```

```python
#Tests for B
print(make_username("Ada", "Lovelace"))
print(make_username("Alan", "Turing"))
print(make_username("Ana Maria", "De Leon"))
```

## C. BOOKEND SWAP PROBLEM

### The function:

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]

user_input = input("Enter items separated by spaces or commas: ")

if "," in user_input:
    items_list = [item.strip() for item in user_input.split(",") if item.strip()]
else:
    items_list = user_input.split()

if len(items_list) >= 2:
    swapped = swap_bookends(items_list)
    print(f"Swapped list: {swapped}")
else:
    print("Please enter at least two items.")
```

```python
#Tests for C
print(swap_bookends([1, 2, 3, 4, 5, 6]))
print(swap_bookends(["red", "green", "blue"]))
print(swap_bookends([8, 3]))
```
